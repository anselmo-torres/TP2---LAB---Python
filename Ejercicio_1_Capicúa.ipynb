```python
#################################Inciso 1#################################
def es_capicua(lista):
    n = len(lista)    #Primero veo el tamaño de la lista
    for i in range(n // 2):   #Defino a i para que corra la lista numero por numero hasta la mitad
        j = n - 1 - i   #j corre la lista pero en reversa
        if lista[i] != lista[j]:    #Entonces si j = i, significa que esa posicion de la lista es capicúa
            return False
    return True


#################################Inciso 2#################################
def es_capicua_plus(lista):
    n = len(lista)
    for i in range(n // 2):
        j = n - 1 - i
        a, b = lista[i], lista[j]
        if a != b and a != -1 and b != -1:    #Si los valores simetricos son distintos, entonces evaluo si alguno de ellos es -1, si no se cumple ya se que la lista no es capicua plus
            return False
    return True


#################################Inciso 3#################################
def es_casi_capicua(lista):
    n = len(lista)
    if n <= 2:
        return True  # quitando 1 elemento queda [] o lista de 0/1 elem, que es capicúa

    for k in range(n):    #Construyo una lista sin la posicion k
        candidata = lista[:k] + lista[k+1:]   #Construyo la lista salteando k. Esto lo hago dividiendo la lista en 2 partes (el medio siendo el elemento k) y despues concateno las 2 partes omitiendo el k
        if es_capicua(candidata):   #Aplica la funcion original a la nueva lista construida
            return True
    return False


#################################Inciso 4#################################
def min_eliminaciones_capicua(lista):
    def MinElim(i, j):
        if i >= j:    #Descartamos las listas que tengan 0 o 1 elementos
            return 0
        if lista[i] == lista[j]:
            return MinElim(i + 1, j - 1)    #Si las puntas coinciden, seguimos hacia adentro
        return 1 + min(MinElim(i + 1, j), MinElim(i, j - 1)) #Si no coinciden, probamos borrar izquierda o derecha y tomamos el mínimo

    return MinElim(0, len(lista) - 1)
```


# Inciso 1: verificación de capicúa

Un lista capicúa es una lista que se puede leer igualmente en un sentido u otro. La forma en que podemos pensar estas listas es que cada elemento de la lista debe ser igual a su "opuesto simétrico", que vendria a ser el elemento cuya posición es la misma solo que contada en el sentido opuesto. En el primer inciso se implementa una función que recorre desde i = 0 hasta i = n//2 - 1 (la mitad de la lista) y compara cada elemento con su opuesto simétrico j, definido como n-1-i, que justamente recorre la lista en el sentido opuesto a i. Si algún i es distinto a j, entonces la función devuelve false, y si i es igual a j para todo i, entonces la función devuelve true.

# Inciso 3: casi capicúa eliminando un elemento

El tercer inciso pregunta si es posible convertir la lista en capicúa eliminando un único elemento. Para ello debeos ver todas las eliminaciones posibles: para cada posición k se construye una lista candidata que contiene todos los elementos excepto el de la posición k. Esa lista candidata se verifica usando la función del inciso 1. Si existe al menos una posición cuyo borrado produce una lista capicúa, la función devuelve True; en caso contrario devuelve False. En listas de longitud muy pequeña, como de tamaño 1 o 2, eliminar un elemento deja 0 o 1 elemento, que generalmente se considera capicúa, por lo que el resultado es directamente verdadero en esos casos.

# Inciso 4: mínimo de eliminaciones para formar una capicúa

El cuarto inciso busca el mínimo número de eliminaciones necesarias para transformar la lista en una capicúa, utilizando recursión. Se define una función recursiva MinElim que representa el mínimo de eliminaciones necesarias para que el subsegmento que va desde el índice i hasta el índice j quede capicúa. El caso base es i >= j, porque un segmento de longitud 0 o 1 ya es capicúa y no requiere eliminaciones; esta condición también cumple el rol de detener la recursión. Si los extremos coinciden, es decir lista[i] == lista[j], no es necesario borrar ninguno de ellos y el problema se reduce al interior, pasando a MinElim(i+1, j-1). Si los extremos no coinciden, no pueden convivir ambos como extremos de una capicúa, por lo que debe eliminarse uno de los dos. En ese caso se consideran dos opciones: borrar el extremo izquierdo y resolver MinElim(i+1, j), o borrar el extremo derecho y resolver MinElim(i, j-1). Como se pide el mínimo, se toma el menor resultado de esas dos posibilidades y se suma uno por el borrado efectuado. La presencia de min en el código se explica porque el problema es de optimización: no alcanza con elegir un lado fijo, ya que en distintas listas conviene eliminar izquierda o derecha dependiendo de la estructura interna
